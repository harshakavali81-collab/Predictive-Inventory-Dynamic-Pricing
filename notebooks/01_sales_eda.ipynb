# 01 | Sales exploratory analysis\nSynthetic-only reproducible example. Run from the repository root.\n

## Synthetic transactions\n

```python
from src.ingestion.generate import generate_sales
from src.processing.validation import validate_sales
from src.analytics.report import summarize
rows = validate_sales(generate_sales())
summary = summarize(rows)
print(summary)
```

## Optional category visualization\n

```python
# Optional: pip install pandas matplotlib
import pandas as pd
import matplotlib.pyplot as plt
df = pd.DataFrame(rows)
df["revenue"] = df["price"] * df["units"]
df.groupby("category")["revenue"].sum().sort_values().plot(kind="barh", title="Synthetic category revenue")
plt.tight_layout()
plt.show()
```
