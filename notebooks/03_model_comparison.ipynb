# 03 | Time-safe forecast comparison\nSynthetic-only reproducible example. Run from the repository root.\n

## Chronological holdout\n

```python
from src.ingestion.generate import generate_sales
from src.forecasting.forecast import backtest
hist = [x["units"] for x in generate_sales() if x["sku"] == "ELEC-101"]
result = backtest(hist)
print("Selected model:", result["selected_model"])
print("Validation MAE:", result["validation_mae"])
print("Sealed holdout:", result["holdout"])
print("Naive baseline:", result["baseline_holdout"])
```

## Holdout plot\n

```python
import matplotlib.pyplot as plt
plt.plot(result["holdout_actual"], label="Actual (synthetic)")
plt.plot(result["holdout_predicted"], label="Predicted")
plt.title("28-day sealed holdout (synthetic)")
plt.legend()
plt.tight_layout()
plt.show()
```
