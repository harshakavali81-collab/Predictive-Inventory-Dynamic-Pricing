# 02 | Weekly sales seasonality\nSynthetic-only reproducible example. Run from the repository root.\n

## Daily synthetic product demand\n

```python
from src.ingestion.generate import generate_sales
rows = [x for x in generate_sales() if x["sku"] == "ELEC-101"]
units = [x["units"] for x in rows]
print("First two weeks", units[:14])
```

## Day-of-week demand pattern\n

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.DataFrame(rows)
df["weekday"] = pd.to_datetime(df["date"]).dt.day_name()
order=["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]
df.groupby("weekday")["units"].mean().reindex(order).plot(kind="bar",title="Synthetic average demand by weekday")
plt.ylabel("Units")
plt.tight_layout()
plt.show()
```
